### **Figura 1.** Participação feminina por gênero literário

![Quinze barras horizontais com a proporção de livros de autoria feminina por gênero literário, de 11% em History a 91% em New Adult; linha tracejada em 51% marca a proporção geral.](https://github.com/andrelfo85/Atividade_Goodreads/blob/master/projeto_goodreads/figuras/fig1_participacao_feminina_por_genero.png?raw=1)



- **Pergunta:** Como a autoria feminina se distribui entre os gêneros
  literários?
- **Por que este gráfico:** A medida comparada é uma proporção (livros de
  autoria feminina sobre o total de livros do gênero) em categorias nominais (os
  gêneros literários). Gráfico de barras horizontais acomodam nomes longos, e a ordenação da menor para a maior proporção facilita a comparação.
- **Como ler:** Cada barra é um gênero literário, segundo a coluna `genero_1`,
  que é o primeiro gênero atribuído ao livro no arquivo CSV contendo o recorte do Goodreads. O comprimento da barra é a proporção de livros de autoria feminina no gênero, e o rótulo ao lado mostra o número de livros (n). A linha tracejada marca a proporção de autoria feminina em toda a E2: 50,5%. A figura exibe os 15 gêneros com mais livros, que somam 7.385 livros (80,4% da E2).
- **O que a figura mostra:** Uma forte separação entre gêneros literários. A
  proporção de autoria feminina vai de 11,3% em History (29 de 257 livros) a

A mediana de avaliações vai de 1.554,5 em Romance (79,0% de autoria feminina) a 77.643 em Classics (19,7%), como mostra a coluna `mediana_avaliacoes` da tabela abaixo. Por isso, uma comparação que junta todos os livros mistura gêneros com níveis de alcance muito diferentes: parte da diferença entre os grupos pode vir dos gêneros em que cada grupo mais publica, e não da autoria em si.

```python
# Tabela equivalente à Figura 1: proporção de livros de autoria feminina
# nos 15 gêneros literários principais (genero_1) com mais livros, com a mediana de avaliações
principais = base_principal["genero_1"].value_counts().head(15).index
livros_principais = base_principal[base_principal["genero_1"].isin(principais)]
tabela_fig1 = (livros_principais.groupby("genero_1")[GRUPO]
               .agg(livros="size", autoria_feminina=lambda s: s.eq("feminino").sum()))
tabela_fig1["proporcao_feminina"] = (tabela_fig1["autoria_feminina"] / tabela_fig1["livros"]).round(3)
tabela_fig1["mediana_avaliacoes"] = livros_principais.groupby("genero_1")["n_avaliacoes"].median()
tabela_fig1 = tabela_fig1.sort_values("proporcao_feminina")

proporcao_geral = base_principal[GRUPO].eq("feminino").mean()
print(f"Proporção geral de autoria feminina na E2 (linha tracejada): {proporcao_geral:.3f}")
print(f"Livros nos 15 gêneros da figura: {tabela_fig1['livros'].sum()} "
      f"({tabela_fig1['livros'].sum() / len(base_principal):.1%} da E2)")
tabela_fig1
```

### **Figura 2.** Distribuição do número de avaliações por livro

![Histogramas em degrau e boxplots do log10 do número de avaliações por grupo; as distribuições se sobrepõem, com a de autoria masculina levemente deslocada para valores maiores.](https://github.com/andrelfo85/Atividade_Goodreads/blob/master/projeto_goodreads/figuras/fig2_distribuicao_avaliacoes_log.png?raw=1)


- **Pergunta:** Como o alcance, medido pelo número de avaliações, se distribui
  entre livros de autoria feminina e masculina?
- **Por que este gráfico:** O número de avaliações é uma variável quantitativa
  muito assimétrica: a maioria dos livros tem poucos milhares de avaliações, e
  alguns têm milhões. O histograma mostra a forma da distribuição; o boxplot
  resume a mediana, os quartis e os valores afastados.
- **Como ler:** O eixo horizontal está em escala logarítmica (log10): 2 = 100
  avaliações; 3 = 1.000; 4 = 10.000; 5 = 100.000; 6 = 1 milhão. Cada unidade a
  mais no eixo multiplica o valor por 10. No painel de cima, a altura de cada
  degrau é a proporção de livros do grupo naquela faixa (faixas de 0,25 em
  log10). No painel de baixo, a caixa vai do percentil 25 ao percentil 75, o
  traço interno é a mediana, as hastes vão até 1,5 vez a amplitude interquartil
  (calculada em log10), e os pontos são livros além das hastes.
- **O que a figura mostra:** As duas distribuições se sobrepõem muito, mas a de
  autoria masculina está deslocada para a direita, isto é, para valores maiores.
  A mediana é de 3.129 avaliações para livros de autoras e de 5.218 para livros
  de autores (razão de 0,60). A segunda tabela abaixo compara cada percentil
  pela razão entre os grupos (autoras / autores): 0,66 no percentil 25; 0,60 na
  mediana; 0,56 no percentil 75; 0,52 nos percentis 90 e 95. Até o percentil
  95, quanto mais avaliado o livro, maior a diferença relativa entre os grupos.
- **Caso:** A relação se inverte ou se atenua nas pontas. No percentil 10,
  livros de autoras têm mais avaliações (202 contra 172). No percentil 99, a
  razão volta a subir para 0,71, e o livro mais avaliado de toda a base é de uma
  autora: *To Kill a Mockingbird*, com 3.819.418 avaliações, contra 3.203.463
  do livro de autoria masculina mais avaliado.
- **Limite:** A escala log permite mostrar no mesmo gráfico livros com 100 e com
  1 milhão de avaliações, mas comprime as diferenças: uma distância pequena no
  eixo pode corresponder a dezenas de milhares de avaliações. As médias (29.506
  e 46.387) não resumem bem esta variável, porque são puxadas por poucos livros
  muito avaliados.

  **===>** A desigualdade de alcance é mais marcada entre os livros muito avaliados (do percentil 75 ao 95) do que no centro, e não aparece nas extremidades da distribuição. Eu faço essa leitura me apoiando apenas nos percentis da tabela.

```python
# Tabela equivalente à Figura 2: distribuição do número de avaliações por grupo
# (contagem, média, desvio-padrão, mínimo, percentis 10, 25, 50, 75 e 90, máximo)
tabela_fig2 = (base_principal.groupby(GRUPO)["n_avaliacoes"]
               .describe(percentiles=[0.1, 0.25, 0.5, 0.75, 0.9])
               .round(0))
display(tabela_fig2)

# Percentis lado a lado e razão entre os grupos (autoras / autores):
# razão menor que 1 indica menos avaliações nos livros de autoras naquele ponto
percentis = [0.10, 0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
tabela_percentis = (base_principal.groupby(GRUPO)["n_avaliacoes"]
                    .quantile(percentis)
                    .unstack(0)
                    .rename_axis("percentil"))
tabela_percentis["razao_feminino_masculino"] = tabela_percentis["feminino"] / tabela_percentis["masculino"]
tabela_percentis.round({"feminino": 1, "masculino": 1, "razao_feminino_masculino": 3})
```

### **Figura 3.** Avaliações e resenhas por livro

![Nuvem de pontos em escala log-log do número de avaliações contra o número de resenhas; triângulos azuis de autoria feminina ficam levemente acima dos círculos laranja de autoria masculina.](https://github.com/andrelfo85/Atividade_Goodreads/blob/master/projeto_goodreads/figuras/fig3_avaliacoes_resenhas.png?raw=1)



- **Pergunta:** O engajamento relativo, isto é, o número de resenhas de um livro em proporção ao número de avaliações que ele recebeu, difere entre livros de autoria feminina e masculina?
- **Por que este gráfico:** São duas variáveis quantitativas medidas no mesmo
  livro (número de avaliações e número de resenhas). O gráfico de dispersão
  mostra como uma varia em relação à outra.
- **Como ler:** Cada ponto é um livro; triângulos azuis são livros de autoras, e
  círculos laranja, livros de autores. Os dois eixos estão em log10. No eixo
  vertical, eu somei 1 ao número de resenhas, porque 11 livros têm zero resenha
  e o logaritmo de zero não existe. Entre dois livros com o mesmo número de
  avaliações, o que está mais alto tem mais resenhas por avaliação.
- **O que a figura mostra:** Avaliações e resenhas crescem juntas (correlação de
  Spearman de 0,94 nos dois grupos). Para comparar o engajamento, a tabela usa a **taxa de resenhas**, que é o número de resenhas dividido pelo número de avaliações (0,110 significa 11 resenhas a cada 100 avaliações). Como essa taxa cai à medida que o livro fica mais avaliado, a comparação é feita dentro de quatro faixas de alcance. Em todas elas, a mediana da taxa é maior para livros de autoras: 0,213 contra 0,179 até mil avaliações; 0,110 contra 0,079 de mil a 10 mil; 0,081 contra 0,056 de 10 mil a 100 mil; 0,053 contra 0,036 acima de 100 mil.
- **Caso:** *The Fault in Our Stars*, de John Green, tem 147.696 resenhas, o
  maior número de toda a base; é o círculo laranja mais alto, no canto superior
  direito. Sua taxa (0,050) está próxima da mediana dos livros de autoras da mesma faixa (0,053) e acima da mediana dos livros de autores (0,036).
- **Limite:** A figura conta resenhas, mas não informa o que elas dizem
  (ex: elogio, crítica). Avaliações também não equivalem a leitores: uma pessoa pode ler um livro sem avaliá-lo.

  **===>** No recorte, livros de autoras recebem mais resenhas em proporção às avaliações, nas quatro faixas de alcance. Isso pode estar ligado à autoria ou aos hábitos das comunidades de leitores de certos gêneros literários.

```python
# Tabela equivalente à Figura 3: mediana da taxa de resenhas por avaliação
# (n_resenhas / n_avaliacoes), por faixa de número de avaliações e grupo
faixas_avaliacoes = pd.cut(base_principal["n_avaliacoes"], [0, 1_000, 10_000, 100_000, 10_000_000],
                           labels=["até 1 mil", "1 mil a 10 mil", "10 mil a 100 mil", "acima de 100 mil"])
tabela_fig3 = (base_principal.assign(faixa_avaliacoes=faixas_avaliacoes)
               .groupby(["faixa_avaliacoes", GRUPO], observed=True)["taxa_resenhas"]
               .agg(livros="size", mediana_taxa="median")
               .round(3)
               .unstack())
display(tabela_fig3)

# Valores auxiliares citados na análise
spearman = (base_principal.groupby(GRUPO)[["n_avaliacoes", "n_resenhas"]]
            .corr(method="spearman").xs("n_resenhas", level=1)["n_avaliacoes"].round(3))
print("\nCorrelação de Spearman entre avaliações e resenhas:", spearman.to_dict())
print("\nLivros com zero resenha:", int(base_principal["n_resenhas"].eq(0).sum()))
print("\nLivros com zero resenha, por grupo:",
      base_principal.loc[base_principal["n_resenhas"].eq(0), GRUPO].value_counts().to_dict())

# Caso citado na minha análise: o livro com mais resenhas no recorte
print("\nLivro com mais resenhas na E2:")
base_principal.nlargest(1, "n_resenhas")[["titulo", "autor", GRUPO, "n_avaliacoes", "n_resenhas", "taxa_resenhas"]]
```

A mediana dos livros de autoras é maior antes de 1990 (10.394 contra 9.030) e de 2010 a 2019 (1.790 contra 1.227), e menor apenas de 1990 a 2009 (6.882 contra 8.104). Como o grupo de autoras concentra livros recentes, sua mediana geral fica abaixo, mesmo sendo maior em dois dos três períodos. E as avaliações foram contadas numa data de coleta desconhecida, de modo que o tempo de exposição de cada livro também é desconhecido.


  **===>** A vantagem dos livros de autores no conjunto da base depende, em grande parte, da composição temporal do recorte. Comparações entre os grupos devem ser feitas dentro de períodos semelhantes.

```python
# Tabela equivalente à Figura 4: número de livros (n) e mediana de avaliações
# por ano de publicação e grupo, de 1990 a 2018
recorte = base_principal[base_principal["ano_publicacao"].between(1990, 2018)]
tabela_fig4 = (recorte.groupby(["ano_publicacao", GRUPO])["n_avaliacoes"]
               .agg(n="size", mediana="median")
               .unstack())

feminina_maior = tabela_fig4[("mediana", "feminino")] > tabela_fig4[("mediana", "masculino")]
print(f"Livros na figura (1990-2018): {len(recorte)} ({len(recorte) / len(base_principal):.1%} da E2)")
print(f"\nAnos de 1990 a 2009 em que a mediana de livros de autoras é maior: {feminina_maior.loc[1990:2009].sum()} de 20")
print(f"\nAnos de 2010 a 2018 em que a mediana de livros de autoras é maior: {feminina_maior.loc[2010:2018].sum()} de 9")

# Composição temporal dos grupos, usada por mim para explicar o caso inesperado
anteriores = base_principal[base_principal["ano_publicacao"] < 1990]
#print(f"\nLivros anteriores a 1990: {len(anteriores)}, em {anteriores['ano_publicacao'].nunique()} anos diferentes "
#      f"(mediana de {anteriores.groupby('ano_publicacao').size().median():.0f} livros por ano)")
print("\nLivros de 2019 por grupo:", base_principal.loc[base_principal["ano_publicacao"] == 2019, GRUPO].value_counts().to_dict())
print("\nAno de publicação mediano:", base_principal.groupby(GRUPO)["ano_publicacao"].median().to_dict())
razao_anual = (tabela_fig4[("mediana", "feminino")] / tabela_fig4[("mediana", "masculino")]).round(2)
print("\nRazão entre medianas (autoras / autores), 2010 a 2018:", razao_anual.loc[2010:2018].to_dict())

# Mediana por período e distribuição de cada grupo entre os períodos (Considerei toda a E2)
periodo = pd.cut(base_principal["ano_publicacao"], [-1000, 1989, 2009, 2019],
                 labels=["antes de 1990", "1990 a 2009", "2010 a 2019"])
por_periodo = (base_principal.assign(periodo=periodo)
               .groupby(["periodo", GRUPO], observed=True)["n_avaliacoes"]
               .agg(livros="size", mediana="median")
               .unstack())
for grupo in ["feminino", "masculino"]:
    por_periodo[("distribuicao", grupo)] = (por_periodo[("livros", grupo)] / por_periodo[("livros", grupo)].sum()).round(3)
print("\nMediana de avaliações, antes de 2010 e de 2010 em diante:",
      base_principal.loc[base_principal["ano_publicacao"] < 2010, "n_avaliacoes"].median(),
      base_principal.loc[base_principal["ano_publicacao"] >= 2010, "n_avaliacoes"].median())
display(tabela_fig4)
por_periodo
```

## 7. Limitações e reprodutibilidade

### Limitações

**Sub-representação:**
+ Literaturas não anglófonas estão sub-representadas.
+ Estão ausentes: livros com menos de cinco avaliações; livros publicados depois de 2019; pessoas não binárias; informações sobre os leitores e sobre o conteúdo das resenhas.

**Escolhas que fiz e que afetam os resultados:**
+ A especificação de gênero (E1, E2, E3) muda o tamanho da diferença de alcance: a razão entre as medianas (autoras / autores) é de 0,71 em E1, 0,60 em E2 e 0,56 em E3.
+ A lista de palavras funcionais retirou *her* e *his*, que poderiam revelar marcas de gênero nos títulos.
+ A tokenização separou contrações.
+ As faixas de ano e de avaliações foram escolhidas por mim, outras faixas poderiam mudar os valores obtidos.
+ A regra de 1,5 IQR foi aplicada em log10.

**Padrões possivelmente produzidos pela construção da base:**
+ Nenhum autor tem mais de seis livros, nenhum livro tem menos de cinco avaliações, e 38,6% dos livros foram publicados entre 2010 e 2014.
+ O erro do rótulo de gênero foi produzido por quem montou o arquivo. Não se pode afirmar se surgiu na coleta ou na montagem do arquivo, porque o método de atribuição do gênero não está documentado.

**Conclusões que não devem ser formuladas:**
+ Que o Goodreads discrimina autoras.
+ Que livros de autoras são melhores ou piores que livros de autores.
+ Que as diferenças observadas valem fora do recorte analisado.
+ Que o gênero inferido pelo nome é o gênero das pessoas.

**Problemas apontados pelo revisor:**

1. A afirmação de que a diferença de alcance entre os grupos “é mais marcada
   entre os livros mais avaliados” não possui uma tabela própria que a
   sustente. Para conferi-la, é preciso comparar, por conta própria,
   os percentis da tabela da `Figura 2`.
2. O relatório usa “autoras” e “autores” como se o grupo fosse o gênero real das
   pessoas. Na verdade, o grupo é o rótulo de gênero registrado no arquivo.
3. A H3 está escrita como afirmação, e não como hipótese provisória.

**Mudanças realizadas após a revisão:**

1. **Sobre o problema 1:** acrescentado à `Figura 2` uma segunda tabela, que põe
   lado a lado os percentis 10, 25, 50, 75, 90, 95 e 99 dos dois grupos e
   calcula a razão entre eles (autoras / autores).
2. **Sobre o problema 2:** inclusão da seguinte nota de terminologia: Neste relatório, “livros de autoras” e “livros de autores” abreviam “livros cuja autoria está registrada como feminina ou masculina no arquivo, segundo a especificação E2”. Não afirmam o gênero de nenhuma pessoa.
3. **Sobre o problema 3:** reescrita da H3 como “Os títulos podem reproduzir
   convenções de gênero literário associadas ao gênero da autoria”.
