## 4. Visualizações

Cada figura vem acompanhada da sua tabela equivalente, gerada pela célula de código logo abaixo da análise. O código repete os cálculos do Notebook 03 desta unidade, onde as figuras são desenhadas, e usa a especificação
principal de gênero da autoria (E2, 9.180 livros).

**Figura 1 e análise:**

![Quinze barras horizontais com a proporção de livros de autoria feminina por gênero literário, de 11% em History a 91% em New Adult; linha tracejada em 51% marca a proporção geral.](https://github.com/andrelfo85/Atividade_Goodreads/blob/master/projeto_goodreads/figuras/fig1_participacao_feminina_por_genero.png?raw=1)

*Pergunta:* Como a autoria feminina se distribui entre gêneros literários?
Barras horizontais ordenadas comparam uma proporção entre categorias. Padrão: segregação forte, de 11% (History, n = 257) a 91% (New Adult, n = 115). Caso: Fantasy, perto do equilíbrio (58%). Limite: composição do recorte, não do mercado. Leitura: qualquer comparação agregada mistura gêneros literários com níveis de alcance muito diferentes.

```python
# Tabela equivalente à Figura 1: proporção de livros de autoria feminina
# nos 15 gêneros literários principais com mais livros
principais = base_principal["genero_1"].value_counts().head(15).index
tabela_fig1 = (base_principal[base_principal["genero_1"].isin(principais)]
               .groupby("genero_1")[GRUPO]
               .agg(livros="size", autoria_feminina=lambda s: s.eq("feminino").sum()))
tabela_fig1["proporcao_feminina"] = (tabela_fig1["autoria_feminina"] / tabela_fig1["livros"]).round(3)
tabela_fig1 = tabela_fig1.sort_values("proporcao_feminina")

proporcao_geral = base_principal[GRUPO].eq("feminino").mean()
print(f"Proporção geral de autoria feminina (linha tracejada da figura): {proporcao_geral:.3f}")
tabela_fig1
```

**Figura 2 e análise:**

![Histogramas em degrau e boxplots do log10 do número de avaliações por grupo; as distribuições se sobrepõem, com a de autoria masculina levemente deslocada para valores maiores.](https://github.com/andrelfo85/Atividade_Goodreads/blob/master/projeto_goodreads/figuras/fig2_distribuicao_avaliacoes_log.png?raw=1)

*Pergunta:* Como o alcance se distribui? Histograma e boxplot em escala log, porque a variável é muito assimétrica. Padrão: grande sobreposição; mediana de 3.129 contra 5.218; percentil 90 de 48.266 contra 92.646. Caso: no percentil 10, livros de autoras têm valor maior (202 contra 172). Limite: a escala log comprime diferenças. Leitura: a desigualdade de alcance é mais marcada entre os livros mais avaliados.



```python
# Tabela equivalente à Figura 2: distribuição do número de avaliações por grupo
# (contagem, média, desvio-padrão, mínimo, percentis 10, 25, 50, 75 e 90, máximo)
tabela_fig2 = (base_principal.groupby(GRUPO)["n_avaliacoes"]
               .describe(percentiles=[0.1, 0.25, 0.5, 0.75, 0.9])
               .round(0))
tabela_fig2
```

**Figura 3 e análise:**

![Nuvem de pontos em escala log-log do número de avaliações contra o número de resenhas; triângulos azuis de autoria feminina ficam levemente acima dos círculos laranja de autoria masculina.](https://github.com/andrelfo85/Atividade_Goodreads/blob/master/projeto_goodreads/figuras/fig3_avaliacoes_resenhas.png?raw=1)

*Pergunta:* O engajamento relativo difere? Dispersão de duas quantitativas
do mesmo livro. Padrão: a taxa de resenhas é maior para livros de autoras em todas as faixas de alcance (0,213 contra 0,179 até mil avaliações; 0,053 contra 0,036 acima de cem mil). Caso: *The Fault in Our Stars* tem 147.696 resenhas, o máximo do recorte. Limite: não informa o teor das resenhas. Leitura: livros de autoras parecem gerar mais escrita por leitor.



```python
# Tabela equivalente à Figura 3: mediana da taxa de resenhas por avaliação
# (n_resenhas / n_avaliacoes), por faixa de número de avaliações e grupo
faixas_avaliacoes = pd.cut(base_principal["n_avaliacoes"], [0, 1_000, 10_000, 100_000, 10_000_000],
                           labels=["até 1 mil", "1 mil a 10 mil", "10 mil a 100 mil", "acima de 100 mil"])
tabela_fig3 = (base_principal.assign(faixa_avaliacoes=faixas_avaliacoes)
               .groupby(["faixa_avaliacoes", GRUPO], observed=True)["taxa_resenhas"]
               .agg(livros="size", mediana_taxa="median")
               .round(3)
               .unstack())
display(tabela_fig3)

# Caso citado na análise: o livro com mais resenhas no recorte
print("Livro com mais resenhas na E2:")
base_principal.nlargest(1, "n_resenhas")[["titulo", "autor", GRUPO, "n_avaliacoes", "n_resenhas"]]
```

**Figura 4 e análise:**

![Duas linhas da mediana de avaliações por ano de publicação entre 1990 e 2018, em escala log; as linhas se cruzam até 2009 e, de 2010 em diante, a de autoria feminina fica sempre acima.](https://github.com/andrelfo85/Atividade_Goodreads/blob/master/projeto_goodreads/figuras/fig4_mediana_por_ano.png?raw=1)

*Pergunta:* A diferença é estável no tempo? Linha, porque o ano é ordenado,
com n anual na tabela. Padrão: de 2010 a 2018 a mediana de livros de autoras é maior em todos os anos (1.599 contra 763 em 2013). Caso inesperado: contraria o resultado agregado. Limite: medianas anuais com poucos casos oscilam (31 livros de autores em 2018). Leitura: a diferença agregada depende da composição temporal do recorte.

```python
# Tabela equivalente à Figura 4: número de livros (n) e mediana de avaliações
# por ano de publicação e grupo, de 1990 a 2018
recorte = base_principal[base_principal["ano_publicacao"].between(1990, 2018)]
tabela_fig4 = (recorte.groupby(["ano_publicacao", GRUPO])["n_avaliacoes"]
               .agg(n="size", mediana="median")
               .unstack())

anos_recentes = tabela_fig4.loc[2010:2018]
maior_feminina = (anos_recentes[("mediana", "feminino")] > anos_recentes[("mediana", "masculino")]).sum()
print(f"Anos de 2010 a 2018 em que a mediana de livros de autoras é maior: {maior_feminina} de {len(anos_recentes)}")
tabela_fig4
```

## 7. Limitações e reprodutibilidade

### Limitações

**Limitações do relatório:**
