### Minha formulação

**Motivação humanística:**
Redes sociais de catalogação e avaliação de livros, como o Goodreads, tornaram-se um espaço em que leitores registram, avaliam e comentam o que leem, e em que a atenção a cada livro fica visível em números. É plausível que livros muito avaliados apareçam mais em listas e recomendações, o que ampliaria diferenças iniciais. Estudar como essa atenção se distribui entre livros de autoras e de autores permite discutir se desigualdades do campo literário se reproduzem, se atenuam ou mudam de forma nesses espaços.

**Conceito central e autores de referência:**
Reconhecimento literário, entendido aqui como atenção pública registrada.

- Merton (1968) descreve o efeito Mateus: quem já é reconhecido tende a receber reconhecimento adicional. Isso ajuda a entender por que contagens de avaliações são tão concentradas em poucos livros.
- Rossiter (1993) nomeia o efeito Matilda: o reconhecimento de mulheres na ciência é sistematicamente subestimado ou atribuído a homens. Eu uso a ideia como hipótese a investigar no campo literário, não como fato demonstrado.
- Thelwall (2017, 2019) analisou o Goodreads e encontrou, em quase todos os gêneros, desequilíbrio na proporção de autoras e autores e, em vários gêneros comuns, avaliações mais altas dadas por leitores a autores do próprio gênero.

**Questão ampla:**
Como o gênero de quem escreve se relaciona com a distribuição do reconhecimento literário em redes sociais de catalogação e avaliação de livros?

**Pergunta delimitada:**

gênero literário principal?

**Finalidade predominante e justificativa:**
Descritiva. O objetivo é medir e descrever a diferença de reconhecimento no recorte. Não pretendo explicar suas causas nem prever o sucesso de livros.

**Estrutura analítica inicial, etapas secundárias e justificativa:**
Estrutura comparativa (autoria feminina × masculina), com uma etapa associativa: examinar se a diferença varia junto com o gênero literário. Etapa secundária: verificar se a composição dos grupos (ano de publicação, gênero
literário) produz parte da diferença.

### Meu mapa de operacionalização

| Conceito | Dimensão | Indicador | Variável | Categorias ou valores | Fonte | Regra | Limitação |
|---|---|---|---|---|---|---|---|
| Reconhecimento na plataforma | Alcance | Número de avaliações | `n_avaliacoes` | inteiro ≥ 5 no recorte | CSV do Goodreads | Comparar medianas e quartis  | Depende do tempo de exposição e da data de coleta; não mede qualidade |

**Por que uma alternativa parece mais adequada à pergunta?**
Para o reconhecimento, nenhuma alternativa substitui a outra: o alcance diz quantas pessoas registraram o livro, e o engajamento diz quantas se dispuseram a escrever sobre ele. A pergunta menciona as duas coisas, então uso as duas como medidas principais. A nota média fica como secundária porque está mais próxima de um julgamento de qualidade. Para o gênero da autoria, o rótulo do arquivo será a medida inicial, e a classificação pelo nome servirá para auditá-lo.

**O que mudou em relação ao mapa do Notebook 02 e por quê?**
Eu incluí a nota média como terceira dimensão, explicitamente secundária, para registrar a decisão de não usá-la como medida principal.

## 5. Evidência e interpretação

Complete a cadeia lógica:

### Minha cadeia de evidência

**Operação prevista:**
Agrupar os livros pelo gênero registrado da autoria e comparar as medianas de `n_avaliacoes` e de `n_resenhas`, no recorte e dentro de cada gênero literário principal com pelo menos 30 livros em cada grupo.

**Resultado possível:**
A mediana de avaliações dos livros de autoria feminina é menor que a dos livros de autoria masculina no recorte, mas a diferença muda de tamanho, ou até de sentido, conforme o gênero literário.

**De que esse resultado poderia ser evidência?**
De que, neste recorte, a atenção registrada pelos usuários se distribui de forma desigual entre os grupos, e de que essa desigualdade depende do gênero literário em que o livro foi classificado.

**O que ele não demonstraria diretamente?**
Que leitores discriminam autoras; Que livros de autoras são piores ou melhores; Que o padrão vale para todo o Goodreads; Ou a causa da diferença.

**Interpretações alternativas:**
- Composição: Autoras podem estar concentradas em gêneros que recebem menos avaliações em geral (por exemplo, romance);
- Tempo de exposição: Livros mais antigos, mais frequentes entre homens no recorte, acumulam avaliações por mais tempo;
- Seleção: O critério de seleção desconhecido do recorte pode ter favorecido um grupo;
- Classificação: Erros no rótulo de gênero podem distorcer os grupos;
- Mercado: Investimento editorial e marketing desiguais, que não estão nos dados.

**Casos ou documentos que precisariam ser relidos:**
Os livros mais avaliados de cada grupo; Livros de autoras em gêneros com predominância masculina e de autores em gêneros com predominância feminina; Casos com rótulo de gênero duvidoso; As páginas desses livros no Goodreads, incluindo resenhas.

**Conhecimento contextual necessário:**
Funcionamento do Goodreads (estantes, estrelas, resenhas, quem usa a rede); História do mercado editorial e dos gêneros literários associados a mulheres; Estudos anteriores sobre gênero e recepção (Thelwall, 2017, 2019).

A cadeia descreve uma interpretação possível. Para não convertê-la em
promessa excessiva, examine perdas, vieses e responsabilidades em todas as
